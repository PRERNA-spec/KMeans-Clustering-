```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import MinMaxScaler
```

```python
da=load_iris()
```

```python
df=pd.DataFrame(da.data,columns=da.feature_names)
```

```python
df
```

```python
df.drop(['sepal width (cm)','petal length (cm)'], axis='columns')
```

```python
plt.scatter(df['sepal length (cm)'],df['petal width (cm)'],color='black')
```

```python
df.isnull().sum()
```

```python
mm=MinMaxScaler()
```

```python
mm.fit_transform(df[['sepal length (cm)']])
```

```python
mm.fit_transform(df[['petal width (cm)']])
```

```python
km=KMeans(n_clusters=3)
km
```

```python
km.fit(df[['sepal length (cm)']],df[['petal width (cm)']])
```

```python
y_pred=km.fit_predict(df[['sepal length (cm)']],df[['petal width (cm)']])
```

```python
y_pred
```

```python
df['clusters']=y_pred
```

```python
df.drop(['sepal width (cm)','petal length (cm)'],axis='columns')
```

```python
df1=df[df.clusters==0]
df2=df[df.clusters==1]
df3=df[df.clusters==2]

plt.scatter(df1['sepal length (cm)'],df1['petal width (cm)'],color='red')
plt.scatter(df2['sepal length (cm)'],df2['petal width (cm)'],color='black')
plt.scatter(df3['sepal length (cm)'],df3['petal width (cm)'],color='blue')
plt.xlabel('sepal length')
plt.ylabel('petal width (cm)')
```

```python
sse=[]
for i in range(1,10):
    km=KMeans(n_clusters=i)
    km.fit(df[['sepal length (cm)']],df[['petal width (cm)']])
    sse.append(km.inertia_)
plt.scatter(range(1,10),sse)    
plt.plot(range(1,10),sse,linestyle='--') 
```
